```python
#PARTE2

import pandas as pd

# Crear columna year
def year_from_title(title):
    try:
        return int(title[-5:-1])
    except:
        return float("nan")

# Quitar el año de title solo si tiene año
def remove_year(title):
    if year_from_title(title) == year_from_title(title):
        return title[:-7]
    else:
        return title

moviesFil["year"] = moviesFil["title"].apply(year_from_title)

# Contar películas sin año reconocible
print("Películas sin año:", moviesFil["year"].isna().sum())

# Asegurarnos de que year sea numérico
moviesFil["year"] = pd.to_numeric(moviesFil["year"], errors="coerce")

# Mostrar algunas películas sin año
print("--------------------------------------")
print(moviesFil.loc[moviesFil["year"].isna(), "title"].head(10))

# Quitar el año de title
moviesFil["title"] = moviesFil["title"].apply(remove_year)

# Mostrar algunas películas sin año
print("--------------------------------------")
print(moviesFil.loc[moviesFil["year"].isna(), "title"].head(10))


```

```python
#PARTE3
#Unimos ratings con movies mediante movieId para añadir el título y los géneros de cada película.
ratings_movies = ratingsFil.merge(
    moviesFil[["movieId", "title", "genres", "year"]],
    on="movieId",
    how="left"
)

print(ratings_movies.head(10))
#También unimos tags con movies mediante movieId:
tags_movies = tagsFil.merge(
    moviesFil[["movieId", "title", "genres", "year"]],
    on="movieId",
    how="left"
)

print(tags_movies.head(10))

# Comprobar el número de filas antes y después del merge
print("Ratings originales:", len(ratingsFil))
print("Ratings después del merge:", len(ratings_movies))

print("Tags originales:", len(tagsFil))
print("Tags después del merge:", len(tags_movies))

# Comprobar si hay movieId duplicados en movies
print("movieId duplicados en movies:", moviesFil["movieId"].duplicated().sum())
```

```python
#PARTE4
# Agrupar por película
peliculas_resumen = ratings_movies.groupby(
    ["movieId", "title"]
).agg(
    num_valoraciones=("rating", "count"),
    media_rating=("rating", "mean"),
    rating_min=("rating", "min"),
    rating_max=("rating", "max")
).reset_index()

print(peliculas_resumen.head(10))


# Crear columna década
ratings_movies["decada"] = (ratings_movies["year"] // 10) * 10
decadas = ratings_movies.groupby("decada").agg(
    num_peliculas=("movieId", "nunique"),
    num_valoraciones=("rating", "count"),
    media_rating=("rating", "mean")
).reset_index()

print(decadas)
ratings_movies["segmento"] = ratings_movies["year"].apply(
    lambda x: "Antes de 2000" if x < 2000 else "2000 o posterior"
)

segmentos = ratings_movies.groupby("segmento").agg(
    num_peliculas=("movieId", "nunique"),
    num_valoraciones=("rating", "count"),
    media_rating=("rating", "mean")
).reset_index()

print(segmentos)
```

```python
#PARTE5
#Cuántas películas están listadas en movies
print("Número de películas:", moviesFil["movieId"].nunique())
print("-----------------------------------------------------")
#Cuáles son las más antiguas (menor año extraído del título).
print(moviesFil.sort_values("year")[["title", "year"]].head(10))
print("-----------------------------------------------------")
#Cuántas tienen "Dracula" en el título (coincidencia parcial, sin distinguir mayúsculas).
dracula = moviesFil["title"].str.contains("Dracula", case=False, na=False)
print("Películas con Dracula:", dracula.sum())
print("-----------------------------------------------------")
#Títulos más comunes.
print(moviesFil["title"].value_counts().head(10))
titulos_repetidos = moviesFil["title"].value_counts()
print(titulos_repetidos[titulos_repetidos > 1])
print("-----------------------------------------------------")



```

```python
#Películas con "Exorcist" ordenadas de la más antigua a la más moderna.
exorcist = moviesFil["title"].str.contains("Exorcist", case=False, na=False)

print(
    moviesFil.loc[exorcist, ["title", "year"]]
    .sort_values("year")
)
print("-----------------------------------------------------")

#Cuántas con año 1950.
print("Películas de 1950:", (moviesFil["year"] == 1950).sum())

print("-----------------------------------------------------")
#Cuántas entre 1950 y 1959 inclusive.
peliculas_50 = moviesFil["year"].between(1950, 1959)

print("Películas entre 1950 y 1959:", peliculas_50.sum())

#Año de la película con título exacto Batman (y contraste con otras Batman).
batman = moviesFil[moviesFil["title"] == "Batman"]

print(batman[["title", "year"]])

batmans = moviesFil[
    moviesFil["title"].str.contains("Batman", case=False, na=False)
]

print(batmans[["title", "year"]].sort_values("year"))


print("-----------------------------------------------------")
```

```python
#Listado de películas que tienen como tag "sci-fi" y "adventure"
tags_sci_fi = tagsFil[tagsFil["tag"].str.lower() == "sci-fi"]
tags_adventure = tagsFil[tagsFil["tag"].str.lower() == "adventure"]
peliculas_sci_fi_adventure = tags_sci_fi[
    tags_sci_fi["movieId"].isin(tags_adventure["movieId"])
]

resultado = peliculas_sci_fi_adventure.merge(
    moviesFil[["movieId", "title"]],
    on="movieId",
    how="left"
)

print(resultado[["movieId", "title"]].drop_duplicates())

print("-----------------------------------------------------")

#¿Cuál es la tag más repetida?
print("Tag más repetida:")
print(tagsFil["tag"].value_counts().head(1))

print(tagsFil["tag"].value_counts().head(10))
```
